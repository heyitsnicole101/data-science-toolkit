```python
# Create a numpy array of size 2 x 2

import numpy as np

array2x2 = np.array([
    [1,2],
    [3,4]
]
)

print("Array size of 2 x 2 :")
print(array2x2)
```

```python
# Create a random numpy array of size 3 x 4

array3x4 = np.random.rand(3,4)

print("Random array of size 3 x 4 :")
print(array3x4)
```

```python
# Create a numpy array of ones and a numpy array of zeroes

array1 = np.ones((3,3))
array0 = np.zeros((3,3))

print("Array of ones: ")
print(array1)

print("Array of zeros: ")
print(array0)
```

```python
# Create a random numpy array of size 6x3 and transpose it

array1 = np.random.rand(6,3)

print("Original Array: ")
print(array1)

array2 = array1.T

print("Transposed Array: ")
print(array2)
```

```python
# Create two random numpy array of size n x p and p x m and then compute the dot product

n = 3
p = 4
m = 5

array1 = np.random.rand(n,p)
array2 = np.random.rand(p,m)

product = np.dot(array1, array2)

print("Array 1:")
print(array1)

print("Array 2:")
print(array2)

print("Dot Product:")
print(product)

print("Result shape:", product.shape)
```
